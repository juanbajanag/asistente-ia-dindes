# 03 — Modelado y arquitectura

El modelo principal del proyecto es una arquitectura RAG. No se realiza fine-tuning del LLM con documentación institucional.


```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent

if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.corpus_builder import build_complete_corpus
from src.cloud_backend import CloudRAGPipeline

records = build_complete_corpus(ROOT / "data", include_sample=True)

print("Raíz del proyecto:", ROOT)
print("Registros:", len(records))
```

```python
pipeline = CloudRAGPipeline(
    dense_min_score=0.45,
    top_k=5,
    temperature=0.0,
).fit(records)

question = "¿Qué componentes principales integra la arquitectura del Simulador Táctico Submarino?"
response = pipeline.ask(question)

print("Modo:", response.response_mode)
print("Abstención:", response.abstained)
print("Fuentes:", response.retrieval_count)
print(response.answer)
```

## Justificación

Se eligió RAG porque:
- mantiene documentos fuera de los pesos del modelo;
- permite actualizar conocimiento sin reentrenamiento;
- facilita mostrar evidencia;
- admite abstención;
- es compatible con operación local/on-premise.
